```python
## OpenAI API Key & Open Source Models & Groq

import os
from dotenv import load_dotenv
load_dotenv()

import openai
openai.api_key = os.getenv("OPENAI_API_KEY")

groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
# from langhain_openai import ChatOpenAI
# calling opensource model without installing it on our system
model = ChatGroq(model="gemma2-9b-it", groq_api_key=groq_api_key)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage(content='Translate the following from English to Spanish. Give answer in Spanish and how to read it in english pronunciation'),
    HumanMessage(content='Heyy Vaishu, will you travel with me ?')
]
res = model.invoke(messages)
```

```python
res
```

```python
# To get only content, i.e. parser
from langchain_core.output_parsers import StrOutputParser
parser = StrOutputParser()
parser.invoke(res)
```

```python
### Using LCEL we can chain components using |; 
# like here are 2 comps, model & parser
chain = model | parser
chain.invoke(messages)
```

```python
### Prompt Templates
from langchain_core.prompts import ChatPromptTemplate

generic_template = 'Translate the following into {language}:'
prompt = ChatPromptTemplate.from_messages(
    [('system',generic_template),
     ('user','{text}')]
)
```

```python
res = prompt.invoke({'language':'French', 'text':'I am sorry'})
```

```python
res.to_messages()
```

```python
chain = prompt | model | parser
chain.invoke({'language':'Spanish','text':'I am sorry'})
```
